```python
import pandas as pd
import numpy as np
from datetime import datetime
```

```python
df = pd.read_csv('churn.csv')
df.head()
```

```python
df.columns
```

```python
df = df.drop(columns=[c for c in df.columns if c.startswith("Unnamed")])
```

```python
df.info()
```

```python
df['region_category'].unique()
```

```python
df['preferred_offer_types'].unique()
```

```python
# Standardizing placeholder junk value to proper NaN
placeholder_value = ['?', 'xxxxxxxx', 'Unknown', 'unknown', '']
df = df.replace(placeholder_value, np.nan)
```

```python
df.info()
```

```python
df.head()
```

```python
df['avg_frequency_login_days'] = pd.to_numeric(df['avg_frequency_login_days'], errors= 'coerce')
```

```python
df['joining_date'] = pd.to_datetime(df['joining_date'], errors = 'coerce')
```

```python
df.info()
```

```python
# Handling missing values
df['region_category'] = df['region_category'].fillna('Unknown')
```

```python
df['joined_through_referral'] = df['joined_through_referral'].fillna('Not Specified')
```

```python
df['preferred_offer_types'] = df['preferred_offer_types'].fillna('No Preference')
```

```python
df['avg_frequency_login_days'] = df['avg_frequency_login_days'].fillna(df['avg_frequency_login_days'].median())
```

```python
df['points_in_wallet'] = df['points_in_wallet'].fillna(df['points_in_wallet'].median())
```

```python
df['medium_of_operation'] = df['medium_of_operation'].fillna('Not Specified')
```

```python
df.info()
```

```python
df.head(20)
```

```python
# Removing invalid values -- value shouldn't be in negative
for col in ['avg_time_spent', 'avg_transaction_value', 'points_in_wallet']:
    before = (df[col] < 0).sum()
    if before > 0:
        print(f' fixing {before} negavtive values in {col}')
    df[col] = df[col].clip(lower=0)
```

```python
# age sanity check (dropping obiviously invalid ages if any exist)
df = df[(df['age'] >=10) & (df['age'] <=100)]
```

```python
# Feature engineering - business-relevant derived fields
# Tenure in days/months since joining (reference date = latest joining_date in data as proxy for "today")
reference_date = df['joining_date'].max()
df['tenure_days'] = (reference_date - df['joining_date']).dt.days
df['tenure_month'] = (df['tenure_days'] / 30).round(1)
```

```python
df.head()
```

```python
# Revenue at risk = transaction value * churn flag
df['revenue_at_risk'] = np.where(
    df['churn_risk_score'] == 1, df['avg_transaction_value'], 0
)
```

```python
df.head()
```

```python
df['avg_frequency_login_days']
```

```python
# Engagement score bucket based on loging frequency
def engagement_bucket(days):
    if pd.isna(days):
        return 'Unknown'
    elif days <= 7:
        return 'Highly Engaged'
    elif days <= 15:
        return 'Moderate Engaged'
    else:
        return 'Low Engaged'

df['engagement_level'] = df['avg_frequency_login_days'].apply(engagement_bucket)
```

```python
df.head()
```

```python
df['membership_category'].unique()
```

```python
# membership tier rank
membership_rank = {
    'No Membership': 0,
    'Basic Membership' : 1,
    'Silver Membership' : 2,
    'Gold Membership' : 3,
    'Premium Membership' : 4,
    'Platinum Membership' : 5
}

df['membership_rank'] = df['membership_category'].map(membership_rank)
```

```python
df.head()
```

```python
df.info()
```

```python
# Complaint flag as boolean
df['has_complaint'] = df['past_complaint'].map({'Yes' : 1, 'No' : 0})
```

```python
df.head()
```

```python
df['feedback'].unique()
```

```python
# feedback category grouped into positive / negative for quick filtering
negative_feedback = [
    'Poor Website', 'Poor Product Quality', 'Poor Customer Service', 'Too many ads'
]
positive_feedback = [
    'Products always in Stock', 'Quality Customer Care', 'User Friendly Website', 'Reasonable Price'
]

def feedback_sentiment(val):
    if val in negative_feedback:
        return 'Negative'
    elif val in positive_feedback:
        return 'Positive'
    else:
        return 'Neutral'

df['feedback_sentiment'] = df['feedback'].apply(feedback_sentiment)
```

```python
df.head()
```

```python
len(df)
```

```python
df.info()
```

```python
# Dropping rows where critical fields are still missing after cleaning
critical_cols = ['age' , 'gender', 'avg_transaction_value' ,'churn_risk_score']
len(df)
df = df.dropna(subset=critical_cols)
len(df)
```

```python
# Removiing duplicate rows by security no which act as unique key
df = df.drop_duplicates(subset=['security_no'])
len(df)
```

```python
output_path = 'cleaned_churn.csv'
df.to_csv(output_path, index=False)
```

```python
print(f"Final shape: {df.shape}")
print(f"\nChurn rate: {df['churn_risk_score'].mean():.1%}")
print(f"\nColumns: {list(df.columns)}")
```
